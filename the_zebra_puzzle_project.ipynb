```python
potters = ['Mummy', 'Daddy', 'Peter', 'Betty']
aunt = 'Aunt'
dreams = ['Sea', 'Ballet', 'Coins', 'Paris']
desserts = ['Napoleon', 'Marmalade', 'Waffles', 'Marshmallows', 'IceCream']
ppl = potters + [aunt]
rules = []
```

```python
# the person is ALWAYS the target
def get_connection(a, b):
  if a in ppl:
        return f"{a}_{b}"
  else:
      return f"{b}_{a}"
# each person MUST connect to an item
def add_required_connection_constraint(kb, target, nodes):
    symbols = [get_connection(target, node) for node in nodes]
    rules.append(f"({' | '.join(symbols)})")
# each person must ONLY connect to a single item
def add_exclusive_connection_constarint(kb, target, nodes):
    symbols = [get_connection(target, node) for node in nodes]
    for i in range(len(nodes)):
      for j in range(i + 1, len(nodes)):
        rules.append(f"~({symbols[i]} & {symbols[j]})")
```

```python
print('Add connection constraints to each person...')
for p in ppl:
    add_required_connection_constraint(rules, p, dreams)
    add_exclusive_connection_constarint(rules, p, dreams)
    add_required_connection_constraint(rules, p, desserts)
    add_exclusive_connection_constarint(rules, p, desserts)
print('Add connection constraints to each dessert...')
for d in desserts:
    add_required_connection_constraint(rules, d, ppl)
print('Add connection constraints to each dream...')
for d in dreams:
    add_required_connection_constraint(rules, d, ppl)
```

```python
# (5) the potters can only dream of: a trip to the sea, a ticket to the ballet, a new collection of coins' album
print("Apply rule (5)...")
for p in potters:
    rules.append(f"~{p}_Paris")
# (2) loving ice-cream -> dreams of visiting paris
print("Apply rule (2)...")
for p in ppl:
    rules.append(f"({p}_IceCream ==> {p}_Paris)")

# (7) Daddy likes the Sea + Peter likes Coins
print("Apply rule (7)...")
rules.append('Daddy_Sea')
rules.append('Peter_Coins')

# (9) Mummy and Daddy have the same dream
print("Apply rule (9)...")
for d in dreams:
    rules.append(f"(Mummy_{d} <=> Daddy_{d})")

# (3) Betty only likes Marmalade
print("Apply rule (3)...")
rules.append('Betty_Marmalade')

# (4) Mummy only likes Marshmallows
print("Apply rule (4)...")
rules.append('Mummy_Marshmallows')

# (8) Peter doesn't like creamy things (Napoleon, IceCream)
print("Apply rule (8)...")
rules.append('~Peter_Napoleon')
rules.append('~Peter_IceCream')

print("Only the Aunt can like ice-cream...")
# only the aunt likes ice-cream
for p in ['Mummy', 'Daddy', 'Peter', 'Betty']:
    rules.append(f"~{p}_IceCream")
```

```python
kb = expr(" & ".join(rules))
```

```python
model = dpll_satisfiable(kb)

if model:
    print("Surviving connestions...")
    for p in ppl:
        p_dream = "Unknown"
        p_dessert = "Unknown"

        for d in dreams:
            if model.get(expr(f"{p}_{d}")) is True:
                p_dream = d
        for d in desserts:
            if model.get(expr(f"{p}_{d}")) is True:
                p_dessert = d
        print(f"{p}: Dreams of {p_dream}, Eats {p_dessert}")
else:
    print("No solution found.")
```
